```python
word = "Python"

print(word[0:3])   # Pyt  (0, 1, 2번 인덱스)
print(word[2:5])   # tho  (2, 3, 4번 인덱스)
print(word[1:4])   # yth  (1, 2, 3번 인덱스)
print("==========") # = 10개 출력
print(word[:3])    # Pyt  (처음부터 2번 인덱스까지)
print(word[3:])    # hon  (3번 인덱스부터 끝까지)
print(word[:])     # Python  (전체)
print("=" * 10)     # = 10개 출력 ( 문자열에 대한 x연산은 반복 )
print(word[::2])    # Pto  (한 칸씩 건너뜀)
print(word[::-1])   # nohtyP  (거꾸로 뒤집기)
```

```python
email = "hong@example.com"

# @ 앞부분(아이디) 추출
at_index = email.index("@") # index 함수 : '@'의 위치 찾기
user_id = email[:at_index]
print("아이디:", user_id)    # 아이디: hong

# 확장자 추출 (뒤에서 3글자)
extension = email[-3:]
print("확장자:", extension)  # 확장자: com
```

```python
sentence = "파이썬 데이터 분석 입문"
words = sentence.split()
print(words)   # ['파이썬', '데이터', '분석', '입문']
print(words[0])
print(words[1])

csv_data = "홍길동,25,서울,데이터분석"
fields = csv_data.split(",")
print(fields)  # ['홍길동', '25', '서울', '데이터분석']
print(fields[0])
print(fields[1])

data = "홍길동:25:서울:개발자"
result = data.split(":", 2)   # 최대 2번만 분리
print(result)   # ['홍길동', '25', '서울:개발자']
```

```python
words = ['파이썬', '데이터', '분석', '입문']

# 공백으로 이어 붙이기
result1 = " ".join(words)
print(result1)   # 파이썬 데이터 분석 입문

# 쉼표로 이어 붙이기
result2 = ",".join(words)
print(result2)   # 파이썬,데이터,분석,입문

# 구분자 없이 이어 붙이기
result3 = "".join(words)
print(result3)   # 파이썬데이터분석입문
```

```python
sentence = "파이썬은 실용적이다. 파이썬은 간결하다."

# '파이썬'을 'Python'으로 모두 바꾸기
result = sentence.replace("파이썬", "Python")
print(result)   # Python은 어렵다. Python은 복잡하다.

# 특정 횟수만 바꾸기 (세 번째 인수)
result2 = sentence.replace("파이썬", "Python", 1)
print(result2)  # Python은 어렵다. 파이썬은 복잡하다.

phone = "010-1234-5678"
clean_phone = phone.replace("-", "") # 제거
print(clean_phone)   # 01012345678
```

```python
name = "홍길동"
age = 25

# 기존 방식 (+ 연산자 사용)
print("이름: " + name + ", 나이: " + str(age))

# f-string 방식 (훨씬 간결!)
print(f"이름: {name}, 나이: {age}")

print("이름: {0}, 나이: {1}".format(name, age))

print("이름: %s, 나이: %d"%(name, age)) # python 2.x, %s:문자열, %d:정수, %f:실수
```

```python
price = 12000
count = 3

print(f"단가: {price}원, 수량: {count}개, 합계: {price * count}원")

pi = 3.141592653589793

print(f"π ≈ {pi:.2f}")   # 소수점 둘째 자리
print(f"π ≈ {pi:.4f}")   # 소수점 넷째 자리

salary = 3500000
print(f"월급: {salary:,}원")   # 월급: 3,500,000원
```

```python
# <: 왼쪽 정렬, >: 오른쪽 정렬, ^: 가운데 정렬
print(f"{'이름':<10}{'점수':>5}")
print(f"{'홍길동':<10}{95:>5}")
print(f"{'김철수':<10}{87:>5}")
print(f"{'이영희':<10}{100:>5}")
```

```python
print("=" * 40) # "="를 40회 반복
print("       연락처 카드 생성기")
print("=" * 40)

# 사용자 입력
full_name = input("이름을 입력하세요: ")
email = input("이메일을 입력하세요: ")
phone = input("전화번호를 입력하세요 (예: 010-1234-5678): ")

# 데이터 가공
name_upper = full_name.upper()          # 이름을 대문자로
email_clean = email.strip().lower()     # 공백 제거 후 소문자로
phone_clean = phone.replace("-", "")    # 전화번호에서 '-' 제거

# 이메일에서 아이디 추출
at_index = email_clean.find("@")
user_id = email_clean[:at_index]

# 결과 출력
print()
print("=" * 40)
print(f"  이름  : {full_name}")
print(f"  영문  : {name_upper}")
print(f"  이메일: {email_clean}")
print(f"  아이디: {user_id}")
print(f"  전화  : {phone_clean[:3]}-{phone_clean[3:7]}-{phone_clean[7:]}")
print("=" * 40)
```
