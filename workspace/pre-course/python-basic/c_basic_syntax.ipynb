```python
# 변수 만들기 ( 변수의 자료형은 저장되는 데이터를 기준으로 결정 )
name = '장동건' # 자료형은 문자열 ( 저장하는 값에 의해 결정 )
age = 45 # 자료형은 정수

print (name, age)

name = '송강호'
age = 52

print (name, age)
```

```python
# 자료형 확인
print( type(name) ) # name 변수의 자료형
print( type(age) ) # age 변수의 자료형
```

```python
# 자료형 변환 ( 형변환 )

# 문자열 → 정수
num_str = "100"
# print(num_str + 5) # 오류 : 문자열과 숫자는 + 할 수 없습니다.
print(num_str + "5")
num_int = int(num_str)
print(num_int + 5)   # 105

# 정수 → 실수
x = 7
print(x)
y = float(x)
print(y)             # 7.0

# 숫자 → 문자열
age = 25
# print("나이: " + age) # 오류 : 문자열과 숫자를 + 연산 할 수 없습니다.
age_str = str(age)
print("나이: " + age_str)  # 나이: 25
```

```python
# 여러 개의 값을 한 번에 출력

name = "홍길동"
age = 25

print(name, age)           # 홍길동 25
print("이름:", name)        # 이름: 홍길동
print("이름:", name, "나이:", age)  # 이름: 홍길동 나이: 25
```

```python
# print의 sep와 end 테스트

# sep 사용: 값 사이를 '-'로 구분
print("2024", "05", "01", sep="-")   # 2024-05-01

# end 사용: 출력 후 줄 바꿈 대신 공백으로 이어 붙이기
print("안녕", end=" ")
print("파이썬!")                       # 안녕 파이썬!
```

```python
# 터미널 환경에서 사용자로부터 키보드 입력 받기

name = input("이름을 입력하세요: ")
print("안녕하세요,", name)
```

```python
# 연산자 연습

a = 10
b = 3

print(a / b)   # 3.3333333333333335
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```
