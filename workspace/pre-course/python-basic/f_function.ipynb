```python
# 함수의 값 반환

def select_random(min, max):
    import random
    n = random.randint(min, max)
    return n # n을 함수를 호출한 곳으로 반환

def show_number():
    n2 = select_random(50, 150) # 함수를 호출하고 함수가 반환한 값을 n에 저장
    print("SELECT NUMBER : ", n2)

show_number()
```

```python
def get_stats(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)
    return total, average, maximum, minimum
```

```python
numbers = [1, 2, 3, 4, 5]
t, a, mx, mn = get_stats(numbers)
print(t, a, mx, mn)

stats = get_stats(numbers)
print(stats)
print(stats[0], stats[1], stats[2], stats[3])
```

```python
# 전달인자의 기본값
def greet(name, greeting="안녕하세요"):
    print(f"{greeting}, {name}님!")

# greet() # 오류 : 전달인자 없이 호출
greet('장동건', '반갑습니다')
greet('송강호') # 기본값이 지정된 전달인자는 호출시 생략 가능합니다.
```

```python
def create_profile(name, age, job):
    print(f"이름: {name}, 나이: {age}, 직업: {job}")

# 위치 인수: 순서대로 전달
create_profile("홍길동", 25, "개발자")

# 키워드 인수: 이름으로 전달 (순서 무관)
create_profile(job="디자이너", name="이영희", age=28)
```

```python
# 가변인자 배열 ( 전달인자의 갯수와 관계 없이 튜플 형태로 수신 )
def total(*args):
    result = 0
    for num in args: # (1, 2, 3)
        result += num
    return result

print(total(1, 2, 3))           # 6
print(total(10, 20, 30, 40))    # 100
print(total(5))                 # 5
```

```python
# 일반 함수
def add(a, b):
    return a + b

add2 = add # 함수는 1급 객체이므로 다른 변수에 저장할 수 있습니다.

# 람다 함수 (동일한 기능)
add_lambda = lambda a, b: a + b

print( add(3, 5) )         # 8
print( add2(3, 5) )         # 8
print( add_lambda(3, 5) )  # 8
```

```python
count = 0   # 전역 변수

def increment():
    global count   # 전역 변수 count를 사용하겠다고 선언
    count += 1

increment()
increment()
increment()
print(count)   # 3
```
